Accuracy= total correct prediction/total prediction
Precision=true positive/true positive+false positive
recall= true positive/true positive+false negative

f1-Score= 2 *(precision * recall)/precision + recall

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics
import seaborn as sns
```

```python
df=pd.read_csv(r"DataSets\customer_churn.csv")
df.head()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='usage_hours', y='customer_churn',marker='o', color='blue',s=100)
plt.title('Customer Churn vs Usage Hours')
plt.xlabel('Usage Hours')
plt.ylabel('Customer Churn')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='subscription_length', y='customer_churn',marker='*', color='red',s=100)
plt.title('Customer Churn vs Usage Hours')
plt.xlabel('Usage Hours')
plt.ylabel('Customer Churn')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='usage_hours', y='customer_churn',marker='X', color='green',s=100)
sns.regplot(data=df, x='usage_hours', y='customer_churn', scatter=False, color='black')
plt.title('Customer Churn vs Usage Hours')
plt.xlabel('Usage Hours')
plt.ylabel('Customer Churn')
plt.show()
```

```python
X=df[['usage_hours','subscription_length']]
y=df['customer_churn']

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)

print(X_train.shape[0])
print(X_test.shape[0])
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(solver='lbfgs')
model.fit(X_train, y_train)


```

```python
y_pred=model.predict(X_test)
print(y_pred)
```

```python
accuracy=model.score(X_test,y_test)
print(accuracy)
print(accuracy*100)
```

```python
proba=model.predict_proba(X_test)
proba_df=pd.DataFrame(proba,columns=['prob(Not Churn)','prob(churn)']).round(2)
proba_df['Actual']=y_test.values
proba_df['Predicted']=y_pred
proba_df
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_pred)
cm
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(cm,annot=True,fmt='d',cmap='Blues',cbar=False,
            xticklabels=['Not Churn','Churn'],
            yticklabels= ['Not Churn','Churn'])
plt.xlabel('Prediated')
plt.ylabel('Actual')
plt.title('confusion_matrix')
plt.show()
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```
