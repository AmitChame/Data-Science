```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import norm 
```

```python
df=pd.read_csv("DataSets/house_prices.csv")
df.head()
```

```python
df.describe()
```

```python
import matplotlib
matplotlib.rcParams['figure.figsize'] = (10, 6)

plt.hist(df['price_lakhs'], bins=20,rwidth=0.8, density=True)
plt.xlabel('Price (in lakhs)')
plt.ylabel('Density')

rng=np.arange(df['price_lakhs'].min(), df['price_lakhs'].max(), 1)
plt.plot(rng, norm.pdf(rng, df['price_lakhs'].mean(), df['price_lakhs'].std()), color='red')
plt.title('House Prices Distribution with Bell Curve')
plt.show()
```

```python
#calculate mean and standard deviation
mean_price = df['price_lakhs'].mean()
std_price = df['price_lakhs'].std()
print(f"Mean Price: {mean_price}, Standard Deviation: {std_price}")
```

```python
#calculate upper and lower limits for outlier detection
upper_limit = mean_price + 3 * std_price # 3 standard deviations above the mean
lower_limit= mean_price - 3 * std_price # 3 standard deviations below the mean
print(f"Upper Limit: {upper_limit}, Lower Limit: {lower_limit}")
```

```python
outliers = df[(df['price_lakhs'] < lower_limit) | (df['price_lakhs'] > upper_limit)]
outliers
```

```python
df_cleaned_std = df[(df['price_lakhs'] >= lower_limit) & (df['price_lakhs'] <= upper_limit)]
df_cleaned_std
```

```python
df_cleaned_std.shape
```

```python
plt.figure(figsize=(10,6))

plt.subplot(1,2,1)
plt.hist(df['price_lakhs'], bins=20,rwidth=0.8, density=True)
plt.title('Original data')
plt.xlabel('Price (in lakhs)')
plt.ylabel('Probability Density')

#cleaned data
plt.subplot(1,2,2)
plt.hist(df_cleaned_std['price_lakhs'], bins=20,rwidth=0.8, density=True)
plt.title('Cleaned data')
plt.xlabel('Price (in lakhs)')
plt.ylabel('Probability Density')

plt.show()
```

Z-Score formula = (X - μ) / σ

```python
df['z_score'] = (df['price_lakhs'] - df['price_lakhs'].mean()) / df['price_lakhs'].std()
print(df.head())
```

```python
outlier_zscore=df[(df['z_score'] < -3) | (df['z_score'] > 3)]
outlier_zscore
```

```python
df_no_outliers= df[(df['z_score'] >= -3) & (df['z_score'] <= 3)]
print(df.shape)
print(df_no_outliers.shape)
```

Exercise: Outlier Detection and Removal using Quantile, Standard Deviation and Z-score Methods


Problem Statement


Use Banglore House Prices (`bhp.csv`) dataset which contains extreme outliers in the `price_per_sqft` column, which can negatively affect machine learning model accuracy.


Your task is to clean this data by removing these outliers.


Tasks to be Performed:
Detect and Remove outliers using the Percentile Method with a [0.001, 0.999] range.
On the cleaned data, detect and remove further outliers using the Standard Deviation Method with a threshold of 4 standard deviations.
Plot a histogram with a bell curve on the final dataset to visualize the cleaned distribution.
Explain how the Z-Score Method is equivalent to the Standard Deviation Method for this task.


```python
df=pd.read_csv("DataSets/bhp.csv")
df.head()
```

```python
df.describe()
```

```python
plt.figure(figsize=(10,6))
plt.hist(df['price_per_sqft'], bins=20,rwidth=0.8, density=True)
plt.xlabel('Price per sqft')
plt.ylabel('Density')
plt.show()
```

```python
#calculate mean and standard deviation
mean_price = df['price_per_sqft'].mean()
std_price = df['price_per_sqft'].std()
print(f"Mean Price per sqft: {mean_price}, Standard Deviation: {std_price}")
```

```python
#calculate upper and lower limits for outlier detection
lower_limit, upper_limit = df['price_per_sqft'].quantile([0.001, 0.999])
print(f"Lower Limit: {lower_limit}, Upper Limit: {upper_limit}")
```

```python
df_percentile_cleaned = df[(df['price_per_sqft'] >= lower_limit) & (df['price_per_sqft'] <= upper_limit)]
df_percentile_cleaned.head()
df.shape, df_percentile_cleaned.shape
```

```python
plt.figure(figsize=(10,6))
plt.hist(df_percentile_cleaned['price_per_sqft'], bins=20,rwidth=0.8, density=True)
plt.xlabel('Price per sqft')
plt.ylabel('Density')
plt.show()
```

```python
df_std=df_percentile_cleaned.copy()
mean_val=df_std['price_per_sqft'].mean()
std_val=df_std['price_per_sqft'].std()
print(f"Mean Price per sqft after cleaning: {mean_val}, Standard Deviation after cleaning: {std_val}")
```

```python
lower_limit_std = mean_val - 4 * std_val
upper_limit_std = mean_val + 4 * std_val
print(f"Lower Limit after cleaning: {lower_limit_std}, Upper Limit after cleaning: {upper_limit_std}")
```

```python
df_std_cleaned = df_std[(df_std['price_per_sqft'] >= lower_limit_std) & (df_std['price_per_sqft'] <= upper_limit_std)]
df_std_cleaned.head()
df_std_cleaned.shape,df_std.shape
```

```python
plt.figure(figsize=(10,6))
plt.subplot(1,2,1)
plt.hist(df_std['price_per_sqft'], bins=20,rwidth=0.8, density=True)
plt.subplot(1,2,2)
plt.hist(df_std_cleaned['price_per_sqft'], bins=20,rwidth=0.8, density=True)
plt.show()
```

```python
#find null values in the dataset
df_std_cleaned.isnull().sum()
```

```python
#z score
df_zscore=df_std_cleaned.copy()
df_zscore['z_score'] = (df_zscore['price_per_sqft'] - df_zscore['price_per_sqft'].mean()) / df_zscore['price_per_sqft'].std()
df_zscore
```

```python
outliers_zscore=df_zscore[(df_zscore['z_score'] < -4) | (df_zscore['z_score'] > 4)]
outliers_zscore.shape
```

```python
df_zscore_cleaned= df_zscore[(df_zscore['z_score'] >= -4) & (df_zscore['z_score'] <= 4)]
df_zscore_cleaned.shape, df_zscore.shape
```

```python
plt.figure(figsize=(10,6))
plt.hist(df_zscore_cleaned['z_score'], bins=20,)
plt.xlabel('Z-Score')
plt.ylabel('Frequency')
plt.title('Distribution of Z-Scores (Cleaned Data)')
plt.show()
```
