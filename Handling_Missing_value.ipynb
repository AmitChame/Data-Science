### Handliing Missing Values

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
# Sample Data with missing values
data = {'A': [1, 2, np.nan, 4, 5],
'B': [np.nan, 2, 3, 4, 5],
'C': [1, 2, 3, np.nan, np.nan],
'D': [10, 20, 30, 40, 50]}
df_missing = pd.DataFrame(data)
print("Original DataFrame with missing values:")
print(df_missing)
```

```python
print(df_missing.isnull().sum())
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df_missing.isnull(), cbar=False, cmap='viridis')
plt.title('Missing Values Heatmap')
plt.show()
```

```python
df_filled_mean = df_missing.copy()

df_filled_mean['A'] = df_filled_mean['A'].fillna(df_filled_mean['A'].mean())
df_filled_mean['B'] = df_filled_mean['B'].fillna(df_filled_mean['B'].mean())
df_filled_mean['C'] = df_filled_mean['C'].fillna(df_filled_mean['C'].mean())

print("DataFrame after filling missing values in column 'A' with mean:")
print(df_filled_mean)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Sample Sales Data
sales_data = {'Product_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
              'Sales_Amount': [150, 200, 120, 300, 180, 250, 100, 400, 220, 170],
              'Category': ['Electronics', 'Clothing', 'Electronics', 'Home Goods', 'Clothing',
                           'Home Goods', 'Electronics', 'Clothing', 'Electronics', 'Home Goods']}
df_sales = pd.DataFrame(sales_data)

print("Original Sales DataFrame:")
print(df_sales)
```

```python
#Histogram of Sales Amount
plt.figure(figsize=(8, 6))
sns.histplot(df_sales['Sales_Amount'], bins=5, kde=True)
plt.title('Distribution of Sales Amount')
plt.xlabel('Sales Amount')
plt.ylabel('Frequency')
plt.show()
```

```python
#Box plot of Sales Amount by Category
plt.figure(figsize=(8, 6))
sns.boxplot(y=df_sales['Sales_Amount'])
plt.title('Box Plot of Sales Amount')
plt.ylabel('Sales Amount')
plt.show()
```

```python
#Bar chart of Sales Amount by Category
plt.figure(figsize=(8, 6))
sns.countplot(x=df_sales['Category'], palette='viridis', hue=df_sales['Category'], legend=False)
plt.title('Count of products per Category')
plt.xlabel('Category')
plt.ylabel('Count')
plt.show()
```

```python
# Add 'Advertising_Spend' and 'Region' for bivariate analysis
df_sales['Advertising_Spend'] = [50, 60, 45, 80, 55, 70, 40, 90, 65, 50]
df_sales['Region'] = ['North', 'South', 'North', 'East', 'West', 'East', 'North', 'South', 'West', 'East']

print("\nSales DataFrame with new columns:")
print(df_sales)
```

```python
#Scatter plot of Sales Amount vs Advertising Spend
plt.figure(figsize=(9, 7))
sns.scatterplot(x='Advertising_Spend', y='Sales_Amount', data=df_sales, hue='Category', size='Sales_Amount', sizes=(20,400))
plt.title('Sales Amount vs Advertising Spend')
plt.xlabel('Advertising Spend') 
plt.ylabel('Sales Amount')
plt.show()
```

```python
# Boxplot of Sales Amount by Category
plt.figure(figsize=(9, 7))
sns.boxplot(x='Category', y='Sales_Amount', data=df_sales, palette='viridis',hue='Category',legend=False)
plt.title('Boxplot of Sales Amount by Category')
plt.xlabel('Category')
plt.ylabel('Sales Amount')
plt.show()
```

```python
# KDE Plot for Sales Amount of Electronics Category
plt.figure(figsize=(9, 7))

sns.kdeplot(
    data=df_sales[df_sales['Category'] == 'Electronics'],
    x='Sales_Amount',
    fill=True,
    color='blue'
)

plt.title('KDE Plot of Sales Amount for Electronics')
plt.xlabel('Sales Amount')
plt.ylabel('Density')
plt.show()
```

```python
plt.figure(figsize=(9, 7))
sns.kdeplot(x='Sales_Amount', data=df_sales, palette='viridis',hue='Category',legend=False,fill=True)
plt.title('KDE Plot of Sales Amount by Category')
plt.xlabel('Sales Amount')
plt.ylabel('Category')
plt.show()
```

```python
# KDE Plot for Sales Amount of Electronics Category
plt.figure(figsize=(9, 7))

sns.kdeplot(
    data=df_sales[df_sales['Category'] == 'Clothing'],
    x='Sales_Amount',
    fill=True,
    color='blue'
)

plt.title('KDE Plot of Sales Amount for Clothing')
plt.xlabel('Sales Amount')
plt.ylabel('Density')
plt.show()
```

```python
# KDE Plot for Sales Amount of Electronics Category
plt.figure(figsize=(9, 7))

sns.kdeplot(
    data=df_sales[df_sales['Category'] == 'Home Goods'],
    x='Sales_Amount',
    fill=True,
    color='blue'
)

plt.title('KDE Plot of Sales Amount for Home Goods')
plt.xlabel('Sales Amount')
plt.ylabel('Density')
plt.show()
```

```python
#create a cross-tabulation table
cross_tab = pd.crosstab(df_sales['Category'], df_sales['Region'])
print("\nCross-tabulation of Category and Region:")
print(cross_tab)
```

```python
#create a cross-tabulation table
cross_tab = pd.crosstab(df_sales['Region'], df_sales['Category'])
print("\nCross-tabulation of Category and Region:")
print(cross_tab)
```

```python
#heatmap for caterical and categorical 
plt.figure(figsize=(8, 6))
sns.heatmap(cross_tab, annot=True, fmt='d', cmap='Blues')
plt.title('Heatmap of Category and Region')
plt.xlabel('Region')
plt.ylabel('Category')
plt.show()
```

**Multivariate Analysis:**

```python
plt.figure(figsize=(10,6))
sns.pairplot(df_sales, vars=['Sales_Amount', 'Advertising_Spend'], hue='Category',palette='viridis',diag_kind='kde')
plt.suptitle('Pairplot of Sales Amount and Advertising Spend by Category', y=1.02) #adjust title position
plt.show()
```

Correlation Heatmap

```python
correlation_matrix = df_sales[['Sales_Amount', 'Advertising_Spend']].corr()
print("\nCorrelation Matrix:")
print(correlation_matrix)
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix Heatmap')
plt.show()
```

Facet Grid 

```python
g=sns.FacetGrid(df_sales, col='Region', row='Category', height=3, aspect=1.2)
g.map(sns.histplot,'Sales_Amount',kde=True,bins=5)
g.set_axis_labels("Sales Amount","Frequency")
g.set_titles(col_template="{col_name} Region",row_template="{row_name} Category")
plt.suptitle('Sales Amount Distribution by Region and Category',y=1.02)
plt.tight_layout(rect=[0,0,1,0.98])
plt.show()
```
