*Seaborn*

```python
#Seaborn is a Python data visualization library based on matplotlib.
#  It provides a high-level interface for drawing attractive and informative statistical graphics.
#  Seaborn is particularly well-suited for visualizing complex datasets and making it easier to understand
#  the relationships between variables.
import numpy as np,pandas as pd,matplotlib.pyplot as plt,seaborn as sns
```

```python
df=pd.read_csv("DataSets/iris.csv")
df.head()
```

```python
corr=df.corr
#correlation matrix
```

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr)
plt.show()
```

```python
#or you can drop the string column
df=df.drop('species',axis=1)
```

```python
corr=df.corr()
sns.heatmap(corr)
```

```python
sns.heatmap(corr, cmap=['red', 'blue','green'],vmin=-.5, vmax=0.6,annot=True)
```

```python
sns.set_theme()
```

```python
tips=sns.load_dataset('tips')
```

```python
#create visualization
sns.relplot(
    data=tips,
    x='total_bill',y='tip',col='time',hue='smoker',size='size',style='smoker',
)
```

Strip Plot

```python
x=['sun','mon','tue','wed','thu','fri','sat']
y=[5,6,7,8,9,10,11]
ax=sns.stripplot(x=x,y=y)
ax.set(xlabel='Days',ylabel='Amount_Spend',title='Stripplot')
plt.show()
```

Boxplot

```python
df=sns.load_dataset('tips')
sns.boxenplot(x='day',y='total_bill',data=df,hue='smoker')
```

Violin plot

```python
# violin plot is similar to box plot, 
# but it also shows the probability density of the data at different values.
sns.violinplot(x='day',y='total_bill',data=df,hue='smoker')
```
