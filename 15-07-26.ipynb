```python


t = np.arange(0, 5, 0.1)

y1 = np.sin(2 * np.pi * t)
y2 = np.cos(2 * np.pi * t)
y3 = np.sin(4 * np.pi * t)
y4 = np.cos(4 * np.pi * t)

plt.figure(figsize=(8, 6))

# Subplot 1
plt.subplot(221)
plt.plot(t, y1, 'b-')
plt.title("sin(2πt)")

# Subplot 2
plt.subplot(222)
plt.plot(t, y2, 'r--')
plt.title("cos(2πt)")

# Subplot 3
plt.subplot(223)
plt.plot(t, y3, 'g-.')
plt.title("sin(4πt)")

# Subplot 4
plt.subplot(224)
plt.plot(t, y4, 'm:')
plt.title("cos(4πt)")


```
