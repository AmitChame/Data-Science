### Setup - Generate Sample Datasets

Run this cell once to create the `data/` folder with sample CSV files used by all four apps
below. If your instructor's official datasets are available, download them and simply overwrite
the files in `data/` with matching column names.

````python
# Setup — create the data folder and sample datasets
# Write your code below

import os
import numpy as np
import pandas as pd

os.makedirs("data", exist_ok=True)

# ---- 1. Retail Sales dataset ----
np.random.seed(42)
dates = pd.date_range("2026-01-01", periods=90, freq="D")
categories = ["Electronics", "Clothing", "Groceries", "Furniture", "Sports"]

rows = []
for d in dates:
    for cat in categories:
        base = {"Electronics": 400, "Clothing": 250, "Groceries": 600,
                "Furniture": 150, "Sports": 200}[cat]
        units = max(0, int(np.random.normal(base / 20, 5)))
        price = {"Electronics": 120, "Clothing": 35, "Groceries": 8,
                 "Furniture": 200, "Sports": 45}[cat]
        revenue = round(units * price * np.random.uniform(0.9, 1.1), 2)
        rows.append([d.date(), cat, units, revenue])

retail_df = pd.DataFrame(rows, columns=["date", "product_category", "units_sold", "revenue"])
retail_df.to_csv("data/retail_sales.csv", index=False)
print("Saved data/retail_sales.csv", retail_df.shape)

# Expected Output: retail_sales.csv created with date, product_category, units_sold, revenue
````

````python
# ---- 2. Student Grades dataset ----
np.random.seed(7)
subjects = ["Mathematics", "Physics", "Chemistry", "English", "Computer Science"]
rows = []
for i in range(1, 151):
    subject = np.random.choice(subjects)
    test1 = np.clip(np.random.normal(65, 12), 0, 100)
    test2 = np.clip(test1 + np.random.normal(0, 8), 0, 100)
    final = np.clip(0.4 * test1 + 0.4 * test2 + np.random.normal(5, 6), 0, 100)
    rows.append([f"S{i:03d}", subject, round(test1, 1), round(test2, 1), round(final, 1)])

grades_df = pd.DataFrame(rows, columns=["student_id", "subject", "test1", "test2", "final_score"])
grades_df.to_csv("data/student_grades.csv", index=False)
print("Saved data/student_grades.csv", grades_df.shape)

# Expected Output: student_grades.csv created with student_id, subject, test1, test2, final_score
````

````python
# ---- 3. Movie Ratings dataset ----
np.random.seed(21)
genres = ["Action", "Comedy", "Drama", "Horror", "Sci-Fi", "Romance"]
rows = []
for i in range(1, 301):
    genre = np.random.choice(genres)
    rating = np.clip(np.random.normal(6.5, 1.2), 1, 10)
    votes = int(np.random.exponential(20000))
    year = int(np.random.randint(1980, 2026))
    title = f"Movie {i:03d}"
    # inject a few missing values in key columns
    if np.random.rand() < 0.03:
        rating = np.nan
    if np.random.rand() < 0.03:
        genre = None
    rows.append([title, genre, round(rating, 1) if pd.notna(rating) else np.nan, votes, year])

movies_df = pd.DataFrame(rows, columns=["title", "genre", "rating", "votes", "year"])
movies_df.to_csv("data/movie_ratings.csv", index=False)
print("Saved data/movie_ratings.csv", movies_df.shape)

# Expected Output: movie_ratings.csv created with title, genre, rating, votes, year
````

````python
# ---- 4. Titanic dataset ----
# Uses the standard, publicly available Titanic dataset
titanic_url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
titanic_df = pd.read_csv(titanic_url)
titanic_df.to_csv("data/titanic.csv", index=False)
print("Saved data/titanic.csv", titanic_df.shape)

# Expected Output: titanic.csv created with survived, pclass, sex, age, etc.
````

### Question 1 - Retail Sales Dashboard

Create an interactive Streamlit web app to analyze daily retail sales data and visualize
product performance.

**Expected Features in the Sales Dashboard WebApp:**
- Load and preview a retail sales dataset.
- Show summary statistics of the mean and median of daily revenue.
- Group and display total units sold and revenue by product category in a table.
- Line chart showing daily revenue trend over time.
- Bar chart comparing total revenue across product categories.

**To run this app:** after executing the cell below, open a terminal in this notebook's folder
and run:
```
streamlit run 1_retail_sales_dashboard.py
```

````python
%%writefile 1_retail_sales_dashboard.py
"""
Experiential Learning Assignment 02 - Question 1
Retail Sales Dashboard
Run with: streamlit run 1_retail_sales_dashboard.py
"""

import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt

st.set_page_config(page_title="Retail Sales Dashboard", layout="wide")
st.title("Retail Sales Dashboard")

# ---- Load and preview the dataset ----
df = pd.read_csv("data/retail_sales.csv", parse_dates=["date"])

st.subheader("Dataset Preview")
st.dataframe(df.head(10))

# ---- Summary statistics of daily revenue ----
st.subheader("Summary Statistics of Daily Revenue")
daily_revenue = df.groupby("date")["revenue"].sum()

col1, col2 = st.columns(2)
col1.metric("Mean Daily Revenue", f"{daily_revenue.mean():,.2f}")
col2.metric("Median Daily Revenue", f"{daily_revenue.median():,.2f}")

# ---- Total units sold and revenue by product category ----
st.subheader("Total Units Sold and Revenue by Product Category")
category_summary = df.groupby("product_category").agg(
    total_units_sold=("units_sold", "sum"),
    total_revenue=("revenue", "sum")
).reset_index().sort_values("total_revenue", ascending=False)

st.dataframe(category_summary)

# ---- Line chart: daily revenue trend over time ----
st.subheader("Daily Revenue Trend Over Time")
fig1, ax1 = plt.subplots(figsize=(10, 4))
ax1.plot(daily_revenue.index, daily_revenue.values, color="teal")
ax1.set_xlabel("Date")
ax1.set_ylabel("Revenue")
ax1.set_title("Daily Revenue Trend")
st.pyplot(fig1)

# ---- Bar chart: total revenue by product category ----
st.subheader("Total Revenue by Product Category")
fig2, ax2 = plt.subplots(figsize=(8, 4))
ax2.bar(category_summary["product_category"], category_summary["total_revenue"], color="coral")
ax2.set_xlabel("Product Category")
ax2.set_ylabel("Total Revenue")
ax2.set_title("Total Revenue by Category")
st.pyplot(fig2)
````

### Question 2 - Student Grades Explorer

Build a Streamlit web app to explore student performance data across different subjects and
visualize their scores.

**Expected Features in the Student Grades Explorer WebApp:**
- Load and preview a grades dataset.
- Allow the user to select a subject from a dropdown menu.
- Display summary statistics for the selected subject: mean, median, and standard deviation of
  final scores.
- Show a boxplot comparing final score distributions across subjects.
- Show a scatterplot of Test1 vs Final scores, colored by subject, for the selected subject.

**To run this app:**
```
streamlit run 2_student_grades_explorer.py
```

````python
%%writefile 2_student_grades_explorer.py
"""
Experiential Learning Assignment 02 - Question 2
Student Grades Explorer
Run with: streamlit run 2_student_grades_explorer.py
"""

import streamlit as st
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

st.set_page_config(page_title="Student Grades Explorer", layout="wide")
st.title("Student Grades Explorer")

# ---- Load and preview the dataset ----
df = pd.read_csv("data/student_grades.csv")

st.subheader("Dataset Preview")
st.dataframe(df.head(10))

# ---- Dropdown to select a subject ----
subjects = sorted(df["subject"].unique())
selected_subject = st.selectbox("Select a Subject", subjects)

subject_df = df[df["subject"] == selected_subject]

# ---- Summary statistics for the selected subject ----
st.subheader(f"Summary Statistics for {selected_subject}")
col1, col2, col3 = st.columns(3)
col1.metric("Mean Final Score", f"{subject_df['final_score'].mean():.2f}")
col2.metric("Median Final Score", f"{subject_df['final_score'].median():.2f}")
col3.metric("Std Dev of Final Score", f"{subject_df['final_score'].std():.2f}")

# ---- Boxplot: final score distributions across subjects ----
st.subheader("Final Score Distribution Across Subjects")
fig1, ax1 = plt.subplots(figsize=(9, 5))
sns.boxplot(data=df, x="subject", y="final_score", ax=ax1)
ax1.set_xlabel("Subject")
ax1.set_ylabel("Final Score")
st.pyplot(fig1)

# ---- Scatterplot: Test1 vs Final Score for the selected subject ----
st.subheader(f"Test1 vs Final Score - {selected_subject}")
fig2, ax2 = plt.subplots(figsize=(8, 5))
sns.scatterplot(data=subject_df, x="test1", y="final_score", hue="subject", ax=ax2)
ax2.set_xlabel("Test1 Score")
ax2.set_ylabel("Final Score")
st.pyplot(fig2)
````

### Question 3 - Movie Ratings Explorer

Develop a Streamlit web app that allows users to explore and analyze movie ratings data based
on selected genres.

**Expected Features in the Movie Ratings Explorer WebApp:**
- Load and preview a movie dataset.
- Drop rows with missing values in key columns.
- Allow the user to select a genre using a dropdown.
- Show summary statistics for the selected genre: average rating, average number of votes, and
  median year of release.
- Show a boxplot comparing the distribution of movie ratings across all genres.
- Display a scatterplot of Votes vs Rating for the selected genre: use vote count as the marker
  size and rating as the color gradient.

**To run this app:**
```
streamlit run 3_movie_ratings_explorer.py
```

````python
%%writefile 3_movie_ratings_explorer.py
"""
Experiential Learning Assignment 02 - Question 3
Movie Ratings Explorer
Run with: streamlit run 3_movie_ratings_explorer.py
"""

import streamlit as st
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

st.set_page_config(page_title="Movie Ratings Explorer", layout="wide")
st.title("Movie Ratings Explorer")

# ---- Load and preview the dataset ----
df = pd.read_csv("data/movie_ratings.csv")

st.subheader("Dataset Preview (Raw)")
st.dataframe(df.head(10))

# ---- Drop rows with missing values in key columns ----
key_columns = ["genre", "rating", "votes", "year"]
df_clean = df.dropna(subset=key_columns)

st.write(f"Rows before cleaning: {len(df)} | Rows after dropping missing key values: {len(df_clean)}")

# ---- Dropdown to select a genre ----
genres = sorted(df_clean["genre"].unique())
selected_genre = st.selectbox("Select a Genre", genres)

genre_df = df_clean[df_clean["genre"] == selected_genre]

# ---- Summary statistics for the selected genre ----
st.subheader(f"Summary Statistics for {selected_genre}")
col1, col2, col3 = st.columns(3)
col1.metric("Average Rating", f"{genre_df['rating'].mean():.2f}")
col2.metric("Average Votes", f"{genre_df['votes'].mean():,.0f}")
col3.metric("Median Release Year", f"{genre_df['year'].median():.0f}")

# ---- Boxplot: rating distribution across all genres ----
st.subheader("Rating Distribution Across All Genres")
fig1, ax1 = plt.subplots(figsize=(9, 5))
sns.boxplot(data=df_clean, x="genre", y="rating", ax=ax1)
ax1.set_xlabel("Genre")
ax1.set_ylabel("Rating")
st.pyplot(fig1)

# ---- Scatterplot: Votes vs Rating for selected genre ----
st.subheader(f"Votes vs Rating - {selected_genre}")
fig2, ax2 = plt.subplots(figsize=(8, 5))
scatter = ax2.scatter(
    genre_df["votes"], genre_df["rating"],
    s=genre_df["votes"] / genre_df["votes"].max() * 300 + 20,
    c=genre_df["rating"], cmap="viridis", alpha=0.7
)
ax2.set_xlabel("Votes")
ax2.set_ylabel("Rating")
fig2.colorbar(scatter, ax=ax2, label="Rating")
st.pyplot(fig2)
````

### Question 4 - Titanic Survival Analysis Dashboard

Build a Streamlit dashboard to analyze survival trends from the Titanic dataset.

**Expected Features in the Titanic Survival Analysis Dashboard WebApp:**
- Preview the dataset and perform basic data cleaning (drop missing values and map numeric
  codes to labels for survival and class).
- Calculate and display survival rates by passenger class and gender using grouped
  percentages.
- Visualize the age distribution by survival status using a box plot to understand age
  patterns among survivors and non-survivors.
- Compute and display the average age for both survivors and non-survivors.

**To run this app:**
```
streamlit run 4_titanic_survival_dashboard.py
```

````python
%%writefile 4_titanic_survival_dashboard.py
"""
Experiential Learning Assignment 02 - Question 4
Titanic Survival Analysis Dashboard
Run with: streamlit run 4_titanic_survival_dashboard.py
"""

import streamlit as st
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

st.set_page_config(page_title="Titanic Survival Analysis", layout="wide")
st.title("Titanic Survival Analysis Dashboard")

# ---- Load the dataset ----
df = pd.read_csv("data/titanic.csv")

st.subheader("Dataset Preview (Raw)")
st.dataframe(df.head(10))

# ---- Basic data cleaning ----
# Drop rows missing key values used in this analysis
df_clean = df.dropna(subset=["age", "survived", "pclass", "sex"]).copy()

# Map numeric codes to labels for survival and class
df_clean["survival_status"] = df_clean["survived"].map({0: "Did Not Survive", 1: "Survived"})
df_clean["class_label"] = df_clean["pclass"].map({1: "First", 2: "Second", 3: "Third"})

st.subheader("Cleaned Dataset Preview")
st.dataframe(df_clean[["survival_status", "class_label", "sex", "age"]].head(10))

# ---- Survival rates by class and gender ----
st.subheader("Survival Rate by Passenger Class and Gender")
survival_rates = (
    df_clean.groupby(["class_label", "sex"])["survived"]
    .mean()
    .mul(100)
    .round(2)
    .reset_index()
    .rename(columns={"survived": "survival_rate_%"})
)
st.dataframe(survival_rates)

# ---- Boxplot: age distribution by survival status ----
st.subheader("Age Distribution by Survival Status")
fig1, ax1 = plt.subplots(figsize=(8, 5))
sns.boxplot(data=df_clean, x="survival_status", y="age", ax=ax1)
ax1.set_xlabel("Survival Status")
ax1.set_ylabel("Age")
st.pyplot(fig1)

# ---- Average age for survivors and non-survivors ----
st.subheader("Average Age: Survivors vs Non-Survivors")
avg_age = df_clean.groupby("survival_status")["age"].mean().round(2)
col1, col2 = st.columns(2)
col1.metric("Avg Age - Survived", f"{avg_age.get('Survived', 0):.2f}")
col2.metric("Avg Age - Did Not Survive", f"{avg_age.get('Did Not Survive', 0):.2f}")
````
