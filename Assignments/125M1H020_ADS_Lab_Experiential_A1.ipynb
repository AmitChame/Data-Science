### Question 1 - Analyze Customer Purchase Patterns

Given a CSV file containing customer transactions (`customer_id`, `purchase_amount`, `date`), use Pandas to calculate the average monthly spending per customer.

```python
# Question 1 — Analyze Customer Purchase Patterns
# Write your code below

import pandas as pd
import numpy as np

# Create a sample transactions CSV file
np.random.seed(42)
n = 60
dates = pd.date_range("2026-01-01", periods=6, freq="MS")

transactions = pd.DataFrame({
    "customer_id": np.random.choice(["C001", "C002", "C003", "C004"], size=n),
    "purchase_amount": np.round(np.random.uniform(100, 2000, size=n), 2),
    "date": np.random.choice(dates, size=n)
})

transactions.to_csv("customer_transactions.csv", index=False)
print("Sample transactions saved to customer_transactions.csv")
transactions.head()

# Expected Output: A CSV file with customer_id, purchase_amount and date columns
```

```python
# Read the CSV file back and extract the month
df = pd.read_csv("customer_transactions.csv", parse_dates=["date"])
df["month"] = df["date"].dt.to_period("M")
df.head()

# Expected Output: DataFrame with an added 'month' column
```

```python
# Calculate average monthly spending per customer
avg_monthly_spending = df.groupby(["customer_id", "month"])["purchase_amount"].mean().reset_index()
avg_monthly_spending.rename(columns={"purchase_amount": "avg_purchase_amount"}, inplace=True)
print(avg_monthly_spending)

# Expected Output: Table of average purchase_amount per customer per month
```

### Question 2 - Identify Missing Values and Impute

Using a dataset with missing values (e.g., age, salary), use Pandas and NumPy to identify missing values and fill them with the column mean.

```python
# Question 2 — Identify Missing Values and Impute
# Write your code below

import pandas as pd
import numpy as np

np.random.seed(1)
data = {
    "name": ["Amit", "Riya", "Karan", "Neha", "Sara", "Vikram", "Puja", "Rohan"],
    "age": [25, np.nan, 32, 28, np.nan, 41, 35, np.nan],
    "salary": [45000, 52000, np.nan, 61000, 58000, np.nan, 47000, 63000]
}
emp_df = pd.DataFrame(data)
print("Original Data:")
emp_df

# Expected Output: DataFrame containing NaN values in 'age' and 'salary'
```

```python
# Identify missing values
print("Missing values per column:")
print(emp_df.isnull().sum())

# Expected Output: Count of missing (NaN) values for each column
```

```python
# Fill missing values with the column mean
emp_df["age"] = emp_df["age"].fillna(emp_df["age"].mean())
emp_df["salary"] = emp_df["salary"].fillna(emp_df["salary"].mean())

print("Data after imputing missing values with column mean:")
emp_df

# Expected Output: DataFrame with no missing values, NaN replaced by column mean
```

### Question 3 - Filter and Sort Employee Records

Given employee data (name, department, salary), use Pandas to filter employees in the "IT" department earning above 60K and sort them by salary.

```python
# Question 3 — Filter and Sort Employee Records
# Write your code below

import pandas as pd

employees = pd.DataFrame({
    "name": ["Amit", "Riya", "Karan", "Neha", "Sara", "Vikram", "Puja", "Rohan"],
    "department": ["IT", "HR", "IT", "Finance", "IT", "IT", "HR", "Finance"],
    "salary": [65000, 48000, 72000, 58000, 55000, 91000, 62000, 47000]
})
employees

# Expected Output: DataFrame with name, department and salary columns
```

```python
# Filter employees in IT department earning above 60K
it_high_earners = employees[(employees["department"] == "IT") & (employees["salary"] > 60000)]

# Sort by salary
it_high_earners_sorted = it_high_earners.sort_values(by="salary", ascending=False)
print(it_high_earners_sorted)

# Expected Output: IT department employees earning above 60000, sorted by salary
```

### Question 4 - Calculate Statistical Metrics for Stock Data

Using NumPy and Pandas, calculate mean, median, and standard deviation of daily stock prices from a dataset.

```python
# Question 4 — Calculate Statistical Metrics for Stock Data
# Write your code below

import pandas as pd
import numpy as np

np.random.seed(7)
stock_prices = pd.DataFrame({
    "date": pd.date_range("2026-07-01", periods=20, freq="D"),
    "close_price": np.round(np.random.normal(loc=150, scale=8, size=20), 2)
})
stock_prices.head()

# Expected Output: DataFrame with date and close_price columns
```

```python
# Calculate mean, median and standard deviation
mean_price = np.mean(stock_prices["close_price"])
median_price = np.median(stock_prices["close_price"])
std_price = np.std(stock_prices["close_price"])

print("Mean Closing Price:", round(mean_price, 2))
print("Median Closing Price:", round(median_price, 2))
print("Standard Deviation:", round(std_price, 2))

# Expected Output: Mean, median and standard deviation of the close_price column
```

### Question 5 - Create a Pivot Table for Sales Analysis

From a sales dataset (region, product, sales), use Pandas to create a pivot table showing total sales per region per product.

```python
# Question 5 — Create a Pivot Table for Sales Analysis
# Write your code below

import pandas as pd

sales_data = pd.DataFrame({
    "region": ["North", "South", "East", "West", "North", "South", "East", "West", "North", "South"],
    "product": ["A", "A", "A", "A", "B", "B", "B", "B", "A", "B"],
    "sales": [12000, 9500, 8700, 11000, 6500, 7200, 8000, 6100, 9800, 7600]
})
sales_data

# Expected Output: DataFrame with region, product and sales columns
```

```python
# Create a pivot table: total sales per region per product
sales_pivot = pd.pivot_table(sales_data, values="sales", index="region",
                              columns="product", aggfunc="sum", fill_value=0)
print(sales_pivot)

# Expected Output: Pivot table with regions as rows, products as columns and total sales as values
```

### Question 6 - Plot Monthly Sales Using Matplotlib

From a dataset with month and sales, use Matplotlib to create a line chart showing monthly sales trends.

```python
# Question 6 — Plot Monthly Sales Using Matplotlib
# Write your code below

import pandas as pd
import matplotlib.pyplot as plt

monthly_sales = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul"],
    "sales": [45000, 52000, 48000, 61000, 58000, 67000, 72000]
})
monthly_sales

# Expected Output: DataFrame with month and sales columns
```

```python
# Plot line chart of monthly sales trend
plt.figure(figsize=(8, 5))
plt.plot(monthly_sales["month"], monthly_sales["sales"], marker="o", color="teal")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()

# Expected Output: Line chart showing sales trend across months
```

### Question 8 - Visualize Correlation Using Heatmap (Seaborn)

Given a dataset with multiple numeric columns, use Seaborn to draw a heatmap of the correlation matrix.

```python
# Question 8 — Visualize Correlation Using Heatmap (Seaborn)
# Write your code below

import pandas as pd
import numpy as np

np.random.seed(3)
numeric_df = pd.DataFrame({
    "age": np.random.randint(20, 60, 50),
    "salary": np.random.randint(30000, 100000, 50),
    "experience": np.random.randint(0, 30, 50),
    "performance_score": np.random.randint(1, 10, 50)
})
numeric_df.head()

# Expected Output: DataFrame with multiple numeric columns
```

```python
# Calculate correlation matrix
corr_matrix = numeric_df.corr()
print(corr_matrix)

# Expected Output: Correlation matrix between all numeric columns
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Draw heatmap of the correlation matrix
plt.figure(figsize=(6, 5))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()

# Expected Output: Heatmap showing correlation values between numeric columns
```

### Question 9 - Visualize Age Distribution Using Histogram

Using a dataset with an age column, plot a histogram with Seaborn to show the age distribution.

```python
# Question 9 — Visualize Age Distribution Using Histogram
# Write your code below

import pandas as pd
import numpy as np

np.random.seed(5)
age_df = pd.DataFrame({
    "age": np.random.normal(loc=35, scale=10, size=200).astype(int)
})
age_df.head()

# Expected Output: DataFrame with an 'age' column
```

### Question 10 - Scatter Plot to Show Relationship Between Two Variables

Use Seaborn to create a scatter plot showing the relationship between advertising budget and sales.

```python
# Question 10 — Scatter Plot to Show Relationship Between Two Variables
# Write your code below

import pandas as pd
import numpy as np

np.random.seed(9)
budget = np.random.uniform(1000, 10000, 40)
sales = budget * 3.5 + np.random.normal(0, 3000, 40)

ad_df = pd.DataFrame({
    "advertising_budget": np.round(budget, 2),
    "sales": np.round(sales, 2)
})
ad_df.head()

# Expected Output: DataFrame with advertising_budget and sales columns
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Scatter plot: advertising budget vs sales
plt.figure(figsize=(7, 5))
sns.scatterplot(data=ad_df, x="advertising_budget", y="sales", color="darkgreen")
plt.title("Advertising Budget vs Sales")
plt.xlabel("Advertising Budget")
plt.ylabel("Sales")
plt.show()

# Expected Output: Scatter plot showing relationship between advertising budget and sales
```
