### Question 2 - Central Tendency: Mean, Median and Mode

Write a Python **function** to compute the **mean**, **median**, and **mode** of a list of numbers.

**Requirements:**
- Define separate functions: `compute_mean(data)`, `compute_median(data)`, `compute_mode(data)`.
- Do **not** use the `statistics` module — implement the logic manually.
- Test your functions with at least two different datasets (one with a unique mode, one with multiple modes).
- Print the result for each measure clearly.

> _Formula reference:_
> - Mean: $\bar{x} = \dfrac{\sum x_i}{n}$
> - Median: Middle value of the sorted list (average of two middle values if $n$ is even)
> - Mode: Value(s) that appear most frequently

```python
# Question 2 — Mean, Median and Mode
# Write your code below


def compute_mean(data):
    return sum(data) / len(data)


def compute_median(data):
    data = sorted(data)
    n = len(data)

    if n % 2 == 0:
        mid1 = data[n // 2 - 1]
        mid2 = data[n // 2]
        return (mid1 + mid2) / 2
    else:
        return data[n // 2]


def compute_mode(data):
    frequency = {}

    for num in data:
        if num in frequency:
            frequency[num] += 1
        else:
            frequency[num] = 1

    max_count = max(frequency.values())

    modes = []

    for key, value in frequency.items():
        if value == max_count:
            modes.append(key)

    if len(modes) == len(frequency):
        return "No Mode"
    elif len(modes) == 1:
        return modes[0]
    else:
        return modes


# Dataset 1 (Unique Mode)
data1 = [10, 20, 20, 30, 40, 50]

print("Dataset 1:", data1)
print("Mean   :", compute_mean(data1))
print("Median :", compute_median(data1))
print("Mode   :", compute_mode(data1))

print()

# Dataset 2 (Multiple Modes)
data2 = [10, 10, 20, 20, 30, 40]

print("Dataset 2:", data2)
print("Mean   :", compute_mean(data2))
print("Median :", compute_median(data2))
print("Mode   :", compute_mode(data2))




# Expected Output:
# Mean   : <value>
# Median : <value>
# Mode   : <value(s)>
```

### Question 3 - Measures of Dispersion: Variance and Standard Deviation

Write a Python **function** to compute the **variance** and **standard deviation** of a list of numbers.

**Requirements:**
- Define functions: `compute_variance(data)` and `compute_std_dev(data)`.
- Implement both **population** and **sample** variants.
- Do **not** use the `statistics` or `numpy` module — implement the logic manually.
- Test your functions with at least two different datasets.

> _Formula reference:_
> - Population Variance: $\sigma^2 = \dfrac{\sum (x_i - \bar{x})^2}{n}$
> - Sample Variance: $s^2 = \dfrac{\sum (x_i - \bar{x})^2}{n - 1}$
> - Standard Deviation: $\sigma = \sqrt{\sigma^2}$

```python
# Question 3 — Variance and Standard Deviation
# Write your code below



def compute_variance(data, sample=False):
    n = len(data)

    mean = sum(data) / n

    total = 0

    for num in data:
        total += (num - mean) ** 2

    if sample:
        return total / (n - 1)
    else:
        return total / n


def compute_std_dev(data, sample=False):
    variance = compute_variance(data, sample)
    return variance ** 0.5


# Dataset 1
data1 = [10, 20, 30, 40, 50]

print("Dataset 1:", data1)
print("Population Variance :", compute_variance(data1))
print("Sample Variance     :", compute_variance(data1, True))
print("Population Std Dev  :", compute_std_dev(data1))
print("Sample Std Dev      :", compute_std_dev(data1, True))

print()

# Dataset 2
data2 = [5, 10, 15, 20, 25, 30]

print("Dataset 2:", data2)
print("Population Variance :", compute_variance(data2))
print("Sample Variance     :", compute_variance(data2, True))
print("Population Std Dev  :", compute_std_dev(data2))
print("Sample Std Dev      :", compute_std_dev(data2, True))



# Expected Output:
# Population Variance : <value>
# Sample Variance     : <value>
# Population Std Dev  : <value>
# Sample Std Dev      : <value>
```

```python
# Question 4(d) — Midpoint of each class interval
# Write your code below

import random
import math

data = [random.randint(1, 100) for i in range(100)]

data_range = max(data) - min(data)

k = math.ceil(1 + 3.322 * math.log10(len(data)))

h = math.ceil(data_range / k)

lower = min(data)

midpoints = []

for i in range(k):
    upper = lower + h - 1
    midpoint = (lower + upper) / 2
    midpoints.append(midpoint)
    lower = upper + 1

print("Class Midpoints")

for m in midpoints:
    print(m)

# Expected Output: List of class midpoints
```

```python
# Question 4(e) — Relative Frequency
# Write your code below

import random
import math

data = [random.randint(1, 100) for i in range(100)]

data_range = max(data) - min(data)

k = math.ceil(1 + 3.322 * math.log10(len(data)))

h = math.ceil(data_range / k)

lower = min(data)

frequencies = []

for i in range(k):
    upper = lower + h - 1

    count = 0

    for value in data:
        if lower <= value <= upper:
            count += 1

    frequencies.append(count)

    lower = upper + 1

relative = []

for f in frequencies:
    relative.append(f / len(data))

print("Relative Frequencies")

for r in relative:
    print(r)

print("Sum =", sum(relative))

# Expected Output: List of relative frequencies (should sum to 1.0)
```

```python
# Question 4(f) — Cumulative Frequency
# Write your code below

import random
import math

data = [random.randint(1, 100) for i in range(100)]

data_range = max(data) - min(data)

k = math.ceil(1 + 3.322 * math.log10(len(data)))

h = math.ceil(data_range / k)

lower = min(data)

frequencies = []

for i in range(k):
    upper = lower + h - 1

    count = 0

    for value in data:
        if lower <= value <= upper:
            count += 1

    frequencies.append(count)

    lower = upper + 1

cumulative = []

total = 0

for f in frequencies:
    total += f
    cumulative.append(total)

print("Cumulative Frequencies")

print(cumulative)
# Expected Output: List of cumulative frequencies (last value should equal n)
```

```python
# Question 4(g) — Complete Frequency Distribution Table (Pandas DataFrame)
# Combine all sub-parts (a) to (f) into one complete table
# Write your code below

import random
import math
import pandas as pd

data = [random.randint(1, 100) for i in range(100)]

data_range = max(data) - min(data)

k = math.ceil(1 + 3.322 * math.log10(len(data)))

h = math.ceil(data_range / k)

lower = min(data)

class_interval = []
frequency = []
midpoint = []
relative_frequency = []
cumulative_frequency = []

total = 0

for i in range(k):

    upper = lower + h - 1

    count = 0

    for value in data:
        if lower <= value <= upper:
            count += 1

    total += count

    class_interval.append(f"{lower}-{upper}")

    frequency.append(count)

    midpoint.append((lower + upper) / 2)

    relative_frequency.append(count / len(data))

    cumulative_frequency.append(total)

    lower = upper + 1

df = pd.DataFrame({
    "Class Interval": class_interval,
    "Frequency": frequency,
    "Midpoint": midpoint,
    "Relative Frequency": relative_frequency,
    "Cumulative Frequency": cumulative_frequency
})

print(df)


# Expected Output: DataFrame with columns:
#   Class Interval | Frequency | Midpoint | Relative Frequency | Cumulative Frequency
```
